```python
import pandas as pd

data = {
    "Name": ["Aman", "Priya", "Rahul", "Sneha", "Karan"],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"],
    "Department": ["IT", "HR", "Finance", "IT", "HR"]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df_label=df.copy()
print(df_label)
df_label["gender-encode"]=le.fit_transform(df_label["Gender"])
print(df_label)
df_label["dep_encodes"]=le.fit_transform(df_label["Department"])
print(df_label)
```

```python
df_encoded=pd.get_dummies(df_label,columns=['City'],dtype="int")
print(df_encoded)
```

```python
import pandas as pd

data = {
    "Age": [18, 22, 25, 30, 35],
    "Salary": [15000, 25000, 40000, 60000, 85000],
    "Height": [150, 160, 170, 175, 180]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler
standard_scaler=StandardScaler();
minmax_scaler=MinMaxScaler();
scaled=standard_scaler.fit_transform(df)
print(scaled)
scaled_df = pd.DataFrame(scaled, columns=df.columns)

print(scaled_df)
```
