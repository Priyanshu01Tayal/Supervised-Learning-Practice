```python
 import pandas as pd

data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 85]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
model=LinearRegression()
X=df[["Hours_Studied"]]
y=df[["Marks"]]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
print(X_train)
print(y_test)
print(X_test)
model.fit(X_train,y_train)
pred=model.predict(X_test)
print(pred)
```

```python
import pandas as pd

data = {
    "Email_Length": [120, 300, 150, 400, 500, 130, 350, 200],
    "Has_Link": [0, 1, 0, 1, 1, 0, 1, 0],
    "Has_Offer_Words": [0, 1, 0, 1, 1, 0, 1, 0],
    "Spam": ["No", "Yes", "No", "Yes", "Yes", "No", "Yes", "No"]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df_encode=df.copy()
print(df_encode)
df["spam_encoded"]=le.fit_transform(df["Spam"])
print(df_encode)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
X=df_encode[['Email_Length','Has_Link','Has_Offer_Words']]
y=df_encode[['spam_encoded']]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
model.fit(X_train,y_train)
pred=model.predict(X_test)[0]
print(pred)
```

```python
import pandas as pd

data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [50, 55, 60, 65, 70, 80, 85, 90],
    "Result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass"]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df_encode=df.copy()
print(df_encode)
df_encode["result"]=le.fit_transform(df_encode["Result"])
print(df_encode)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
X=df_encode[['Hours_Studied','Attendance']]
y=df_encode['result']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
model = KNeighborsClassifier(n_neighbors=3)

model.fit(X_train, y_train)

pred = model.predict(X_test)

print(pred)
```

```python
import pandas as pd

data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "Attendance": [50, 55, 60, 65, 70, 75, 80, 90],
    "Assignments_Completed": [1, 1, 2, 2, 3, 3, 4, 5],
    "Result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass", "Pass", "Pass"]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
le=LabelEncoder()
df_encode=df.copy()
df_encode['result_encode']=le.fit_transform(df_encode['Result'])
print(df_encode)
X=df_encode[['Hours_Studied','Attendance','Assignments_Completed']]
y=df_encode['result_encode']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
de=DecisionTreeClassifier();
de.fit(X_train,y_train)
ans=de.predict(X_test)
print(ans)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

de = DecisionTreeClassifier()

de.fit(X_train, y_train)

ans = de.predict(X_test)

print(ans)

accuracy = accuracy_score(y_test, ans)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, ans))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, ans)

print(cm)
```
